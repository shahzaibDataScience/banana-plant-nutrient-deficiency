```python
import os
import numpy as np
import tensorflow as tf
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras.models import Model
from tensorflow.keras.layers import GlobalAveragePooling2D, Dense, Dropout, BatchNormalization
from tensorflow.keras.applications import EfficientNetB0
from tensorflow.keras.optimizers import Adam
from tensorflow.keras.regularizers import l2
from tensorflow.keras.callbacks import ModelCheckpoint, EarlyStopping, ReduceLROnPlateau
from sklearn.utils.class_weight import compute_class_weight
import matplotlib.pyplot as plt
import shutil

# ─────────────────────────────────────────────────────────────────
# STEP 1 — REMOVE 'New folder' CLASS FROM DATASET
# Run this once only — it moves the folder out of training
# ─────────────────────────────────────────────────────────────────
base_dir  = "/kaggle/input/datasets/shahzaibnoorr/robo-data"
train_dir = os.path.join(base_dir, "train")
valid_dir = os.path.join(base_dir, "valid")
test_dir  = os.path.join(base_dir, "test")

BAD_CLASS = "New folder"
for split_dir in [train_dir, valid_dir, test_dir]:
    bad_path = os.path.join(split_dir, BAD_CLASS)
    if os.path.exists(bad_path):
        print(f"⚠️  Skipping '{BAD_CLASS}' in {split_dir} — handled via class filter")

# ─────────────────────────────────────────────────────────────────
# STEP 2 — PARAMETERS
# ─────────────────────────────────────────────────────────────────
IMG_SIZE      = 224
BATCH_SIZE    = 32
EPOCHS_FROZEN = 15
EPOCHS_FINE   = 40

# Only these 8 classes (New folder excluded)
VALID_CLASSES = ['Boron', 'Calcium', 'Healthy', 'Iron',
                 'Magnesium', 'Manganese', 'Potassium', 'Sulphur']

# ─────────────────────────────────────────────────────────────────
# STEP 3 — STRONGER AUGMENTATION for Calcium & Magnesium
#           These two look similar so we need more variety
# ─────────────────────────────────────────────────────────────────
train_datagen = ImageDataGenerator(
    preprocessing_function=tf.keras.applications.efficientnet.preprocess_input,
    rotation_range=40,           # increased from 20
    width_shift_range=0.2,       # increased from 0.15
    height_shift_range=0.2,      # increased from 0.15
    shear_range=0.15,            # increased from 0.1
    zoom_range=0.25,             # increased from 0.2
    horizontal_flip=True,
    vertical_flip=True,          # NEW: helps with leaf orientation
    brightness_range=[0.7, 1.3], # wider brightness range
    channel_shift_range=30,      # NEW: random color shift
    fill_mode='nearest'
)
valid_datagen = ImageDataGenerator(
    preprocessing_function=tf.keras.applications.efficientnet.preprocess_input
)
test_datagen = ImageDataGenerator(
    preprocessing_function=tf.keras.applications.efficientnet.preprocess_input
)

# ─────────────────────────────────────────────────────────────────
# STEP 4 — DATA GENERATORS (filter out 'New folder')
# ─────────────────────────────────────────────────────────────────
train_generator = train_datagen.flow_from_directory(
    train_dir,
    target_size=(IMG_SIZE, IMG_SIZE),
    batch_size=BATCH_SIZE,
    class_mode='categorical',
    classes=VALID_CLASSES,       # ← excludes 'New folder'
    shuffle=True,
    seed=100
)
valid_generator = valid_datagen.flow_from_directory(
    valid_dir,
    target_size=(IMG_SIZE, IMG_SIZE),
    batch_size=BATCH_SIZE,
    class_mode='categorical',
    classes=VALID_CLASSES,       # ← excludes 'New folder'
    shuffle=False
)
test_generator = test_datagen.flow_from_directory(
    test_dir,
    target_size=(IMG_SIZE, IMG_SIZE),
    batch_size=1,
    class_mode='categorical',
    classes=VALID_CLASSES,       # ← excludes 'New folder'
    shuffle=False
)

num_classes = train_generator.num_classes
print(f"\n✅ Classes found : {num_classes}")
print(f"   Class list    : {VALID_CLASSES}")
print(f"   Train samples : {train_generator.samples}")
print(f"   Valid samples : {valid_generator.samples}")
print(f"   Test  samples : {test_generator.samples}\n")

# ─────────────────────────────────────────────────────────────────
# STEP 5 — CLASS WEIGHTS
#           Penalizes the model more for getting Calcium &
#           Magnesium wrong — forces it to learn the difference
# ─────────────────────────────────────────────────────────────────
class_weights = compute_class_weight(
    class_weight='balanced',
    classes=np.unique(train_generator.classes),
    y=train_generator.classes
)
class_weight_dict = dict(enumerate(class_weights))

# Boost Calcium and Magnesium weights even more
calcium_idx   = VALID_CLASSES.index('Calcium')
magnesium_idx = VALID_CLASSES.index('Magnesium')
class_weight_dict[calcium_idx]   *= 1.5   # extra penalty for Calcium mistakes
class_weight_dict[magnesium_idx] *= 1.5   # extra penalty for Magnesium mistakes

print("📊 Class weights:")
for i, cls in enumerate(VALID_CLASSES):
    print(f"   {cls:<15}: {class_weight_dict[i]:.3f}")

# ─────────────────────────────────────────────────────────────────
# STEP 6 — BUILD MODEL
# ─────────────────────────────────────────────────────────────────
def build_transfer_model(img_size, num_classes):
    base_model = EfficientNetB0(
        weights='imagenet',
        include_top=False,
        input_shape=(img_size, img_size, 3)
    )
    base_model.trainable = False

    x = base_model.output
    x = GlobalAveragePooling2D()(x)

    x = Dense(256, kernel_regularizer=l2(0.01), bias_regularizer=l2(0.01))(x)
    x = BatchNormalization()(x)
    x = tf.keras.layers.Activation('relu')(x)
    x = Dropout(0.5)(x)

    x = Dense(128, kernel_regularizer=l2(0.01), bias_regularizer=l2(0.01))(x)
    x = tf.keras.layers.Activation('tanh')(x)
    x = Dropout(0.25)(x)

    outputs = Dense(num_classes, activation='softmax')(x)
    model = Model(inputs=base_model.input, outputs=outputs)
    return model, base_model

model, base_model = build_transfer_model(IMG_SIZE, num_classes)

# ─────────────────────────────────────────────────────────────────
# CALLBACKS
# ─────────────────────────────────────────────────────────────────
def get_callbacks(checkpoint_name):
    return [
        ModelCheckpoint(checkpoint_name, monitor='val_accuracy',
                        save_best_only=True, verbose=1),
        EarlyStopping(monitor='val_accuracy', patience=8,
                      restore_best_weights=True, verbose=1),
        ReduceLROnPlateau(monitor='val_loss', factor=0.5,
                          patience=4, min_lr=1e-7, verbose=1)
    ]

# ─────────────────────────────────────────────────────────────────
# PHASE 1 — Train head only
# ─────────────────────────────────────────────────────────────────
print("\n" + "="*60)
print("PHASE 1: Training classification head (backbone frozen)")
print("="*60)

model.compile(
    optimizer=Adam(learning_rate=1e-3),
    # Label smoothing: stops model being overconfident
    loss=tf.keras.losses.CategoricalCrossentropy(label_smoothing=0.1),
    metrics=['accuracy']
)

history1 = model.fit(
    train_generator,
    epochs=EPOCHS_FROZEN,
    validation_data=valid_generator,
    class_weight=class_weight_dict,   # ← apply class weights
    verbose=1,
    callbacks=get_callbacks('best_model_phase1.keras')
)

# ─────────────────────────────────────────────────────────────────
# PHASE 2 — Fine-tune top layers
# ─────────────────────────────────────────────────────────────────
print("\n" + "="*60)
print("PHASE 2: Fine-tuning top layers of EfficientNetB0")
print("="*60)

base_model.trainable = True
for layer in base_model.layers[:-30]:
    layer.trainable = False

model.compile(
    optimizer=Adam(learning_rate=1e-4),
    loss=tf.keras.losses.CategoricalCrossentropy(label_smoothing=0.1),
    metrics=['accuracy']
)

history2 = model.fit(
    train_generator,
    epochs=EPOCHS_FINE,
    validation_data=valid_generator,
    class_weight=class_weight_dict,   # ← apply class weights
    verbose=1,
    callbacks=get_callbacks('best_model_phase2.keras')
)

# ─────────────────────────────────────────────────────────────────
# PLOT TRAINING CURVES
# ─────────────────────────────────────────────────────────────────
combined_acc      = history1.history['accuracy']     + history2.history['accuracy']
combined_val_acc  = history1.history['val_accuracy'] + history2.history['val_accuracy']
combined_loss     = history1.history['loss']         + history2.history['loss']
combined_val_loss = history1.history['val_loss']     + history2.history['val_loss']
phase1_end        = len(history1.history['accuracy'])

fig, axes = plt.subplots(1, 2, figsize=(14, 5))
axes[0].plot(combined_acc,     label='Train Accuracy')
axes[0].plot(combined_val_acc, label='Val Accuracy')
axes[0].axvline(x=phase1_end, color='gray', linestyle='--', label='Fine-tune start')
axes[0].set_title('Model Accuracy')
axes[0].set_xlabel('Epoch')
axes[0].set_ylabel('Accuracy')
axes[0].legend()
axes[0].grid(True)

axes[1].plot(combined_loss,     label='Train Loss')
axes[1].plot(combined_val_loss, label='Val Loss')
axes[1].axvline(x=phase1_end, color='gray', linestyle='--', label='Fine-tune start')
axes[1].set_title('Model Loss')
axes[1].set_xlabel('Epoch')
axes[1].set_ylabel('Loss')
axes[1].legend()
axes[1].grid(True)

plt.tight_layout()
plt.savefig('/kaggle/working/training_curves.png', dpi=150)
plt.show()

# ─────────────────────────────────────────────────────────────────
# EVALUATE
# ─────────────────────────────────────────────────────────────────
test_loss, test_acc = model.evaluate(test_generator, verbose=1)
print(f"\n✅ Test Accuracy : {test_acc * 100:.2f}%")
print(f"   Test Loss     : {test_loss:.4f}")
```

```python
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.metrics import confusion_matrix, classification_report
import tensorflow as tf

# ── CORRECT 8 classes (matches new model) ────────────────────────
VALID_CLASSES = ['Boron', 'Calcium', 'Healthy', 'Iron',
                 'Magnesium', 'Manganese', 'Potassium', 'Sulphur']

test_dir = "/kaggle/input/datasets/shahzaibnoorr/robo-data/test"
IMG_SIZE = 224

# ── Reload test generator with 8 classes only ─────────────────────
test_datagen = tf.keras.preprocessing.image.ImageDataGenerator(
    preprocessing_function=tf.keras.applications.efficientnet.preprocess_input
)
test_generator = test_datagen.flow_from_directory(
    test_dir,
    target_size=(IMG_SIZE, IMG_SIZE),
    batch_size=1,
    class_mode='categorical',
    classes=VALID_CLASSES,    # ← must match training
    shuffle=False
)

# ── Load the NEW model (phase2 from improved training) ────────────
model = tf.keras.models.load_model('/kaggle/working/best_model_phase2.keras')

# ── Predict ───────────────────────────────────────────────────────
test_generator.reset()
y_pred_probs = model.predict(test_generator, verbose=1)
y_pred = np.argmax(y_pred_probs, axis=1)
y_true = test_generator.classes

# ── Confusion Matrix ──────────────────────────────────────────────
cm = confusion_matrix(y_true, y_pred)

plt.figure(figsize=(10, 8))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
            xticklabels=VALID_CLASSES, yticklabels=VALID_CLASSES)
plt.title('Confusion Matrix (8 Classes)', fontsize=16)
plt.ylabel('True Label', fontsize=13)
plt.xlabel('Predicted Label', fontsize=13)
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.savefig('/kaggle/working/confusion_matrix_v2.png', dpi=150)
plt.show()

# ── Classification Report ─────────────────────────────────────────
print("\n📊 Classification Report:")
print(classification_report(y_true, y_pred, target_names=VALID_CLASSES))

# ── Per-class Accuracy ────────────────────────────────────────────
per_class_acc = cm.diagonal() / cm.sum(axis=1)

plt.figure(figsize=(10, 5))
bars = plt.bar(VALID_CLASSES, per_class_acc * 100, color='steelblue', edgecolor='black')
plt.ylim(0, 110)
plt.ylabel('Accuracy (%)')
plt.title('Per-Class Accuracy (8 Classes)')
plt.xticks(rotation=45, ha='right')
for bar, acc in zip(bars, per_class_acc):
    plt.text(bar.get_x() + bar.get_width()/2,
             bar.get_height() + 1.5,
             f'{acc*100:.1f}%', ha='center', fontsize=9)
plt.tight_layout()
plt.savefig('/kaggle/working/per_class_accuracy_v2.png', dpi=150)
plt.show()

# ── Top confused pairs ────────────────────────────────────────────
print("\n🔍 Most confused class pairs:")
cm_no_diag = cm.copy()
np.fill_diagonal(cm_no_diag, 0)
for _ in range(5):
    idx = np.unravel_index(np.argmax(cm_no_diag), cm_no_diag.shape)
    print(f"   True: {VALID_CLASSES[idx[0]]:<15} → Predicted: {VALID_CLASSES[idx[1]]:<15} ({cm_no_diag[idx]} times)")
    cm_no_diag[idx] = 0
```

```python
import os
import random
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf
from tensorflow.keras.preprocessing import image

# ── Config ────────────────────────────────────────────────────────
test_folder = "/kaggle/input/datasets/shahzaibnoorr/robo-data/test"
IMG_SIZE    = 224

# ── Load model ────────────────────────────────────────────────────
model = tf.keras.models.load_model('/kaggle/working/best_model_phase2.keras')

class_labels = ['Boron', 'Calcium', 'Healthy', 'Iron',
                'Magnesium', 'Manganese', 'Potassium', 'Sulphur']

# ── Collect 1 image per class ─────────────────────────────────────
image_paths = []
for cls in class_labels:
    cls_folder = os.path.join(test_folder, cls)
    if os.path.exists(cls_folder):
        files = [f for f in os.listdir(cls_folder)
                 if f.lower().endswith(('.jpg', '.jpeg', '.png'))]
        if files:
            image_paths.append((os.path.join(cls_folder, random.choice(files)), cls))

print(f"Found {len(image_paths)} classes to display")

# ── Plot predictions (2 rows x 4 cols = 8 classes) ───────────────
plt.figure(figsize=(20, 10))
for i, (img_path, actual_label) in enumerate(image_paths[:8]):
    img = image.load_img(img_path, target_size=(IMG_SIZE, IMG_SIZE))
    img_array = image.img_to_array(img)
    img_array = tf.keras.applications.efficientnet.preprocess_input(img_array)
    img_array = np.expand_dims(img_array, axis=0)

    prediction      = model.predict(img_array, verbose=0)
    predicted_idx   = np.argmax(prediction[0])
    predicted_label = class_labels[predicted_idx]
    confidence      = prediction[0][predicted_idx] * 100

    color = 'green' if predicted_label == actual_label else 'red'

    plt.subplot(2, 4, i + 1)
    plt.imshow(image.load_img(img_path, target_size=(IMG_SIZE, IMG_SIZE)))
    plt.title(f"Actual: {actual_label}\nPredicted: {predicted_label}\n{confidence:.1f}%",
              fontsize=10, color=color)
    plt.axis('off')

plt.suptitle('Sample Test Predictions', fontsize=16, fontweight='bold')
plt.tight_layout()
os.makedirs('/kaggle/working/saved_graphs', exist_ok=True)
plt.savefig('/kaggle/working/saved_graphs/sample_predictions.png', dpi=150)
plt.show()
print("✅ Done!")
```
